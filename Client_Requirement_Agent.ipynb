```python
!pip install google-generativeai

from google.colab import userdata

import google.generativeai as genai

api_key = userdata.get('Gemini_API_Key_2')

genai.configure(api_key=api_key)

model = genai.GenerativeModel("models/gemini-3.6-flash")
```

```python
def project_requirement_agent(project_type, business_name, budget, deadline, details):
    prompt = f"""
    Tum ek professional web development requirement-gathering assistant ho.
    Client ne ye details di hain:

    Project Type: {project_type}
    Business Name: {business_name}
    Budget: {budget}
    Deadline: {deadline}
    Extra Details: {details}

    Ab ek clear Project Requirement Summary banao jisme ye shamil ho:
    - Project Overview
    - Key Features Needed
    - Suggested Tech Stack (WordPress/Shopify/Custom)
    - Estimated Timeline
    - Next Steps
    """
    response = model.generate_content(prompt)
    return response.text
```

```python
result = project_requirement_agent(
    "Shopify Store",
    "22inc.shop",
    "PKR 50,000",
    "3 weeks",
    "Luxury menswear brand, dark/gold theme, need product pages and checkout"
)
print(result)
```

P (Performance): Complete aur clear project requirements gather karna

E (Environment): Client ke diye gaye answers (project type, budget, deadline, details)

A (Actuators): Structured requirement summary generate karna

S (Sensors): Client ka typed input
